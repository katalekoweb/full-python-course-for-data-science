### Lists

```python
# objects that can change the values and can have duplicated values
countries = ["Angola", "Angola", "Benguela", "Brasil", "Portugal"]
```

```python
countries
```

```python
# the index starts at 0
print(countries[0])
print(countries[1])
print(countries[2])
print(countries[3])
```

```python
# get the last element of the list
print(countries[-1])
print(countries[-2])
```

```python
# slicing
print(countries[0:3])
print(countries[2:5])
print(countries[1:])
print(countries[:2])
```

```python
# add a country in end of list
countries.append("Mozambique")
```

```python
print(countries)
```

```python
# add a country in the start of list
countries.insert(0, "Zambia")
```

```python
print(countries)
```

```python
# join list
countries_2 = ["UK", "Germany", "Austria"]
countries_2
```

```python
countries + countries_2
```

```python
nested_list = [countries, countries_2]
```

```python
nested_list
```

```python
# remove with item value
countries.remove('Angola')
```

```python
countries
```

```python
# remove the last element
countries.pop()
```

```python
countries
```

```python
# delete from index
del countries[0]
```

```python
countries
```

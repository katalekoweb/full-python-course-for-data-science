### Option 2 - create a Dataframe from a dictionary

```python
# lists used for this example
states = ['Benguela', 'Navegantes', 'Lobito', 'Cubal']
population = [800030, 750000, 800000, 300008]
```

```python
# creating random integer number between 1 to 100
int_language_score = np.random.randint(1, 100, size=1000)
```

```python
# create random foat number
# np.random.uniform(1,100,size=1000)
```

```python
# creating random numbers for our new score columns
score1 = np.random.randint(1,100,size=1000)
score2 = np.random.randint(1,100,size=1000)
```

```python
# crate series from the random numbers
series1 = pd.Series(score1, index=np.arange(0,1000))
series2 = pd.Series(score2, index=np.arange(0,1000))
```

```python
# using assign to add a new column
df_exams = df_exams.assign(score1=series1,score2=series2)
```

```python
# show a dataframe
df_exams
```

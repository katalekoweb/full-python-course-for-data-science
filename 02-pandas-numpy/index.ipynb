### Option 1 using Array

```python
# creating an array
data = np.array([[1,4], [2,5], [3,6]])
```

```python
# creating a dataframe
df = pd.DataFrame(data, index=['row1', 'row2', 'row3'], columns=['col1', 'col2'])
```

```python
# showing the dataframe
df
```

### Option 2 - create a Dataframe from a dictionary

```python
# lists used for this example
states = ['Benguela', 'Navegantes', 'Lobito', 'Cubal']
population = [800030, 750000, 800000, 300008]
```

```python
# Storing lists with a dictionary
dict_states = {'states': states, 'population': population}
```

```python
# creating the dataframe
df_population = pd.DataFrame(dict_states)
```

```python
# showing the databaframe
df_population
```

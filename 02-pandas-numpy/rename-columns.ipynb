```python
import pandas as pd
```

## 1. Rename Columns

```python
# reading file
df_exams = pd.read_csv('StudentsPerformance.csv')
```

```python
# rename column and overwrite of the dataframe
df_exams = df_exams.rename(columns={'gender': 'Gender'})
```

```python
# rename multiple columns and update the dataframe with inplace arguments
df_exams.rename(columns={
    'math score': 'MS', 
    'reading score': 'RS',
    'wring score': 'WS'
}, inplace=True)
```

```python
# show dataframe
df_exams
```

## 2. Rename index

```python
# rename index 0, 1 and 2 and update the dataframe
df_exams.rename(index={0:'A',1:'B',2:'C'}, inplace=True)
```

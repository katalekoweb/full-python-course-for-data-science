```python
import pandas as pd
import numpy as np
import random
```

```python
# reading file
df_exams = pd.read_csv('StudentsPerformance.csv')
```

```python
# sort by one column
df_exams.sort_values(by='math score')
```

```python
# sort descending by one column
df_exams.sort_values(by='math score', ascending=False)
```

```python
# sort by multiple columns
df_exams.sort_values(by=['math score', 'reading score'])
```

```python
# sort descending by multiple columns
df_exams.sort_values(by=['math score', 'reading score'], ascending=False)
```

```python
# sort descending by multiple columns and update the dataframe
df_exams.sort_values(by=['math score', 'reading score'], ascending=False, inplace=True)
```

```python
df_exams
```

```python
# sort with text using a key function
df_exams.sort_values('race/ethnicity', ascending=True, key=lambda col:col.str.lower())
```

## 1 Create Index

```python
# creating non repetitive values for the index
new_index = np.arange(0, 1000)
new_index
```

```python
# shuffling the indexes
random.shuffle(new_index)
```

```python
new_index
```

```python
# create a new column with the new index
df_exams['new_index'] = new_index
```

```python
df_exams
```

```python
# set a new index to the dataframe
df_exams.set_index('new_index', inplace=True)
```

```python
# sort by index
df_exams.sort_index()
```

```python
# descending sort by index
df_exams.sort_index(ascending=False, inplace=True)
df_exams
```

## 2. Read .csv from multiple URLs with Pandas

```python
# https://www.football-data.co.uk/mmz4281/2122/E0.csv
```

```python
# link structure
"https://www.football-data.co.uk/mmz4281/" + "2122" + "/" + "E0" + ".csv"
```

```python
# create a root variable
root = "https://www.football-data.co.uk/mmz4281/"
```

```python
### 2.1 Multiple Leagues
```

```python
# creating a list of leagues
leagues = ['E0', 'E1', 'E2', 'E3', 'EC']

frames = []

# looping through league, read multiple csv and append into a list
for league in leagues:
    df = pd.read_csv(root + '2122' + '/' + league + '.csv', encoding='unicode_escape')
    frames.append(df)
```

```python
# length of frames
len(frames)
```

```python
# show 1st, 2nd and 3rd elements
frames[0]
```

### 2. Multiple seasoes and leagues

```python
# creating a list of leagues
leagues = ['E0', 'E2', 'E3']

frames = []

# looping through league, read multiple csv and append into a list
for league in leagues:
    for season in range(15,21):
        df = pd.read_csv(root + str(season) + str(season+1) + '/' + league + '.csv')
        df.insert(1, 'season', season)
        frames.append(df)
```

```python
# total frames (6 seasons x 3 leagues)
len(frames)
```

```python
# show first frame e last frame
frames[0]
frames[17]
```

### 3. Organize all the data in a dictionary

```python
# create a dictionary with original name of league key
dict_countries = {
    'Spanish La Liga': 'SP1',
    'Spanish Segunda Division': 'SP2',
    'German Bundesliga': 'D1',
    'English Premier League': 'E0'
}
```

```python
# get a dictionary element
dict_countries['Spanish La Liga']
```

```python
dict_historical_data = {}

# creating a list of leagues
# leagues = ['E0', 'E2', 'E3']


# looping through league, read multiple csv and append into a list
for league in dict_countries:
    frames = []
    for season in range(15,21):
        df = pd.read_csv(root + str(season) + str(season+1) + '/' + dict_countries[league] + '.csv')
        df.insert(1, 'season', season)
        frames.append(df)
    df_concat = pd.concat(frames)
    dict_historical_data[league] = df_concat
```
